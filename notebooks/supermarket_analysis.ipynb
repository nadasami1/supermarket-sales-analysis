```python
df.drop_duplicates(inplace=True)
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python
print(df['Date'].dtype)
```

```python
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
```

```python
print(df['Sales'].dtype)
```

```python
sales_null=((df['Unit price']*df['Quantity'])+df['Tax 5%'])
```

```python
df.fillna({'Sales':sales_null},inplace=True)
```

```python
df['Sales'].isna().sum()
```

```python
df.isna().sum()
```

```python
UnitPrice_null= (df['Sales']-df['Tax 5%'])/df['Quantity']
```

```python
df.fillna({'Unit price':UnitPrice_null},inplace=True)
```

```python
df['Unit price'].isna().sum()
```

```python
df['Year']=df['Date'].dt.year
```

```python
df['Month']=df['Date'].dt.month
```

```python
df['Day']=df['Date'].dt.day
```

```python
df.head()
```

```python
df['Branch']=df['Branch'].str.replace("giza","Giza")
df['Branch']=df['Branch'].str.replace("cairo","Cairo")
```

```python
df['Satisfaction'] = df['Rating'].apply(lambda Rating: 'Satisfied' if Rating>=7 else 'Not Satisfied' )
```

```python
df['Satisfaction']
```

```python
df.head()
```

```python
df.to_csv('SuperMarket_CleanedData')
```

## Data Analysis

```python
data = pd.DataFrame(pd.read_csv('SuperMarket_CleanedData'))
```

```python
data.columns
```

```python
Total_Revenue = data['Sales'].sum().round(2)
print(f'TotalRevenue is : {Total_Revenue}')
```

```python
print(f'TotalRevenue is : {Total_Revenue}')
```

```python
Cities_Revenue = data.groupby('City')['Sales'].sum().round(2).sort_values(ascending=False).reset_index()
Cities_Revenue
```

```python
print(f' the city with the highest revenue is : {Cities_Revenue.iloc[0,0]}')
```

```python
Branches_Profit = data.groupby('Branch')['gross income'].sum().round(2).sort_values(ascending=False).reset_index()
Branches_Profit
```

```python
print(f' the Branch with the highest Profit is : {Branches_Profit.iloc[0,0]}')
```

```python
Category_Performance = data.groupby('Product line')[['Sales','gross income']].sum().round(2).sort_values(['Sales','gross income'],ascending=[False,False]).reset_index()
Category_Performance
```

```python
print(f'the top-performing category is : {Category_Performance.iloc[0,0]}')
```

```python
Customers_Style = data.groupby('Customer type')['Sales'].sum().sort_values(ascending=False).reset_index()
Customers_Style
```

```python
print(f'the customer type spends more is : {Customers_Style.iloc[0,0]}')
```

```python
Payment_Methods = data.groupby('Payment')['Invoice ID'].count().sort_values(ascending=False).reset_index()
Payment_Methods
```

```python
print(f'the most popular payment method is : {Payment_Methods.iloc[0,0]}')
```

```python
Average_Transaction = data['Sales'].sum().mean()
```

```python
print(f'the average transaction value for the supermarket is : {Average_Transaction}')
```

```python
Customer_Satisfaction = data.groupby('Branch')['Satisfaction'].value_counts(normalize=True).mul(100).round(2).sort_values(ascending=False).reset_index(name='Percentage')
Customer_Satisfaction
```

```python
highest_Satisfaction =Customer_Satisfaction[Customer_Satisfaction['Satisfaction'] == 'Satisfied'].sort_values('Percentage',ascending=False)
highest_Satisfaction
```

```python
print(f'The highest_Satisfaction Branch is : {highest_Satisfaction.iloc[0,0]} with a {highest_Satisfaction.iloc[0]['Percentage']} % Percentage')
```

```python
Revenue_trend = data.groupby(['Year','Day','Month'])['Sales'].sum().round(2).sort_values(ascending=False).reset_index()
Revenue_trend
```

```python
print(f'The highest Sales was in :\n{Revenue_trend.iloc[0,0:3]}')
```
